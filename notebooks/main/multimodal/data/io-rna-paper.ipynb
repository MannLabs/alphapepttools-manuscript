```python
import anndata as ad
import pandas as pd
```

```python
metadata = pd.read_csv(
    "./raw/metadata.tsv", sep="\t", index_col=0
).drop(columns=["tmt_set_and_label"])
adata = ad.io.read_excel(
    "./raw/lethioe2021_SI1.xlsx", sheet="Table2-Transcriptomics"
).T
adata.X = adata.X.astype(float)
adata.obs = adata.obs.merge(metadata, left_index=True, right_index=True)
```

```python
adata.write_h5ad("./pp/karlsson2014_nslc_rna.h5ad")
```
