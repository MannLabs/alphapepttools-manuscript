1. Lehtiö, J. et al. Proteogenomics of non-small cell lung cancer reveals molecular subtypes associated with specific therapeutic targets and immune-evasion mechanisms. Nat Cancer 2, 1224–1242 (2021).

See Supplementary File 1


```python
import anndata as ad
import pandas as pd
```

```python
expression_data = (
    pd.read_excel(
        "./raw/metadata/lethioe2021_SI1.xlsx", sheet_name="Table2-Mehylation"
    )
    .set_index(["GeneSymbol", "probeID", "ID"])
    .T
)

adata = ad.AnnData(
    X=expression_data.values,
    obs=expression_data.index.to_frame(index=False, name="sample_id").set_index(
        "sample_id"
    ),
    var=expression_data.columns.to_frame(index=False).set_index("GeneSymbol"),
)
adata.X = adata.X.astype(float)
```

```python
metadata = pd.read_csv(
    "https://datashare.biochem.mpg.de/s/x5gMNoZnagZAonT/download", sep="\t", index_col=0
).drop(columns=["tmt_set_and_label"])
adata.obs = adata.obs.merge(metadata, left_index=True, right_index=True)
```

```python
adata.write_h5ad("./pp/karlsson2014_nslc_methylation.h5ad")
```
